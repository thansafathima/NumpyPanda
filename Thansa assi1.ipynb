## PART 3: CREATING NEW COLUMNS & CALCULATIONS (PANDAS + NUMPY)

### 16. Create a new column Revenue as Sales - Discount.

```python
df["Revenue"] = df["Sales"] - df["Discount"]
```

### 17. Create a new column Profit Ratio as Profit / Sales.

```python
df["Profit Ratio"] = df["Profit"] / df["Sales"]
```

### 18. Create a new column High Sales Flag using NumPy (True if Sales > 1000 else False).

```python
df["High Sales Flag"] = np.where(df["Sales"] > 1000, True, False)
```

### 19. Calculate total sales per Region using a Pandas column operation.

```python
sales_per_region = df.groupby("Region")["Sales"].transform("sum")
```

### 20. Calculate average profit per Category using a column operation.

```python
avg_profit_per_category = df.groupby("Category")["Profit"].transform("mean")
```

### 21. Using NumPy, create a new column Sales Log as the natural log of Sales.

```python
df["Sales Log"] = np.log(df["Sales"])
```

### 22. Using NumPy, create a new column Profit Squared as Profit squared.

```python
df["Profit Squared"] = np.square(df["Profit"])
```

## PART 4: AGGREGATION AND GROUPBY (PANDAS)

### 23. Find total sales per Region.

```python
total_sales_region = df.groupby("Region")["Sales"].sum()
print("\n--- Total Sales per Region ---")
print(total_sales_region)
```

### 24. Find average profit per Category.

```python
avg_profit_category = df.groupby("Category")["Profit"].mean()
print("\n--- Average Profit per Category ---")
print(avg_profit_category)
```

### 25. Find maximum and minimum sales per Sub-Category.

```python
sales_min_max = df.groupby("Sub-Category")["Sales"].agg(["min", "max"])
print("\n--- Min & Max Sales per Sub-Category ---")
print(sales_min_max)
```

### 26. Count the number of orders per Customer Segment.

```python
orders_per_segment = df.groupby("Segment")["Order ID"].nunique()
print("\n--- Order Count per Segment ---")
print(orders_per_segment)
```

### 27. Find total quantity sold per Product Name and list top 10 products.

```python
top_10_quantity_products = (
    df.groupby("Product Name")["Quantity"].sum().nlargest(10)
)
print("\n--- Top 10 Products by Total Quantity Sold ---")
print(top_10_quantity_products)
```

### 28. Find total sales and total profit for each Category per Region.

```python
cat_region_summary = df.groupby(["Region", "Category"])[["Sales", "Profit"]].sum()
print("\n--- Total Sales & Profit per Category & Region ---")
print(cat_region_summary)
```

### 29. Find average discount per Sub-Category per Region.

```python
avg_discount_subcat_region = df.groupby(["Region", "Sub-Category"])["Discount"].mean()
print("\n--- Avg Discount per Sub-Category per Region ---")
print(avg_discount_subcat_region)
```

### 30. Count the number of orders per Category per Segment.

```python
orders_cat_segment = df.groupby(["Segment", "Category"])["Order ID"].nunique()
print("\n--- Orders per Category per Segment ---")
print(orders_cat_segment)
```

## PART 5: NUMPY ARRAY TASKS

### 31. Convert Sales column to a NumPy array and find statistics.

```python
sales_arr = df["Sales"].to_numpy()
print("\n--- NumPy Sales Array Statistics ---")
print(f"Maximum: {np.max(sales_arr):.2f}")
print(f"Minimum: {np.min(sales_arr):.2f}")
print(f"Mean: {np.mean(sales_arr):.2f}")
print(f"Median: {np.median(sales_arr):.2f}")
print(f"Standard Deviation: {np.std(sales_arr):.2f}")
```

### 32. Using NumPy, find all orders where Profit > 500.

```python
profit_arr = df["Profit"].to_numpy()
high_profit_indices = np.where(profit_arr > 500)[0]
high_profit_orders_np = df.iloc[high_profit_indices]
```

### 33. Using NumPy, create a normalized Sales column using (Sales - mean) / std.

```python
sales_mean = np.mean(sales_arr)
sales_std = np.std(sales_arr)
df["Sales Normalized"] = (sales_arr - sales_mean) / sales_std
```

### 34. Using NumPy, calculate the sum of Revenue for orders where Profit Ratio > 0.2.

```python
revenue_arr = df["Revenue"].to_numpy()
profit_ratio_arr = df["Profit Ratio"].to_numpy()
high_ratio_revenue_sum = np.sum(revenue_arr[profit_ratio_arr > 0.2])
print(f"\nSum of Revenue (Profit Ratio > 0.2): {high_ratio_revenue_sum:.2f}")
```

### 35. Using NumPy, find the indices of the top 5 highest Sales orders.

```python
top_5_sales_indices = np.argsort(sales_arr)[-5:][::-1]
print("\n--- Top 5 Sales Indices ---")
print(top_5_sales_indices)
print("\n--- Top 5 Highest Sales Orders ---")
print(df.iloc[top_5_sales_indices][["Order ID", "Sales"]])
```
