## 5. Mathematical Operations – Business Metrics

```python
#37. Increase product prices by multiplying with tax factor 1.1
base_prices = np.array([1000, 2000, 3000])
q37 = base_prices * 1.1
print(q37)
```

```python
#38. Square of customer purchase quantities
qty = np.array([2, 3, 4, 5])
q38 = qty ** 2
print(q38)
```

```python
#39. Calculate profit by subtracting cost from revenue
rev = np.array([5000, 6000, 7000])
cost = np.array([3000, 3500, 4000])
q39 = rev - cost
print(q39)
```

```python
#40. Calculate discounted prices by dividing by 2
orig_prices = np.array([2000, 4000, 6000])
q40 = orig_prices / 2
print(q40)
```

```python
#41. Find the total daily revenue
daily_rev = np.array([12000, 15000, 18000, 20000])
q41 = np.sum(daily_rev)
print(q41)
```

```python
#42. Find maximum and minimum product prices
prod_prices = np.array([450, 1200, 700, 300, 900])
q42_max = np.max(prod_prices)
q42_min = np.min(prod_prices)
print("Max:", q42_max, "| Min:", q42_min)
```

## 6. Statistical Operations – Data Analytics 

```python
#43. Average daily sales
sales_data = np.array([12000, 15000, 18000, 20000, 17000])
q43 = np.mean(sales_data)
print(q43)
```

```python
#44. Median customer purchase value
purchase_vals = np.array([500, 700, 900, 1200, 1500])
q44 = np.median(purchase_vals)
print(q44)
```

```python
#45. Variance of monthly revenue
monthly_rev = np.array([10000, 12000, 14000, 16000])
q45 = np.var(monthly_rev)
print(q45)
```

```python
#46. Standard deviation of product ratings
ratings = np.array([3, 4, 5, 4, 5])
q46 = np.std(ratings)
print(q46)
```
