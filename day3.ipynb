## BOX PLOT

```python

import matplotlib.pyplot as plt
```

```python
data = [10,12,14,15,18,20,22,25,30,100]
plt.boxplot(data,vert=False)
plt.title("Before Removing Outliers")
plt.show()
```

```python
import numpy as np
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
print(Q1)
print(Q3)
IQR=Q3-Q1
print(IQR)
```

```python
lower_limit=Q1-(1.5*IQR)
upper_limit=Q3+(1.5*IQR)
print("Lower limit:",lower_limit)
print("Upper limit:",upper_limit)
```

```python
outliers=[x for x in data if x<lower_limit or x>upper_limit]
print(outliers)
```

```python
clean_data=[x for x in data if lower_limit<= x <= upper_limit]
clean_data
```

## LOAD DATASET

```python
df = sns.load_dataset('tips')
df.head()
```

## BASIC PLOTS

```python
sns.lineplot(x='total_bill', y='tip', data=df)
plt.title('Line Plot')
plt.show()
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=df)
plt.title('Scatter Plot')
plt.show()
```

```python
sns.barplot(x='day', y='total_bill', data=df)
plt.title('Bar Plot')
plt.show()
```

```python
sns.barplot(x='day', y='total_bill',data=df, ci=None)
plt.title('Bar plot')
plt.show()
```

## Distribution PLOts

```python
sns.histplot(df['total_bill'], kde=True)
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=df)
plt.show()
```

## Heat Map

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True,cmap = "coolwarm")
plt.show()
```

## TIME SERIES PLOT

```python
dates = [datetime.datetime(2024,1,i) for i in range(1,6)]
values = [10,20,15,25,30]
print(dates)
plt.plot(dates, values)
plt.xticks(dates,rotation=45)
plt.show()
```
