```python
import pandas as pd

data = {
    "Department": ["HR", "IT", "IT", "HR", "Finance", "IT", "Finance"],
    "Employee": ["A", "B", "C", "D", "E", "F", "G"],
    "Salary": [40000, 60000, 65000, 45000, 70000, 62000, 72000],
    "Experience": [2, 5, 6, 3, 8, 4, 9],
    "City":["Bangalore","Chennai","Kochi","Kochi","Chennai","Kochi","Mumbai"]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")['Salary'].max()
```

```python
df.groupby("Department")['Salary'].min()
```

```python
df.groupby('Department')['Employee'].count()
```

```python
df.groupby('City')['Employee'].count()
```

```python
##AGGREGATE USING agg()
```

```python
#average, maximum and minimum salary in each department
df.groupby("Department")["Salary"].agg(["mean","max","min"])
```

```python
#maximum and minimum experience in each department
df.groupby("Department")['Experience'].agg(['max','min'])
```

```python
##AGGREGATE USING agg()multiple columns
```

```python
df.groupby("Department").agg({
    "Salary": ["mean","max"],
    "Experience": ["min","max"]
})
```

##Named Aggregation

```python
df.groupby("Department").agg(
    Average_Salary=("Salary","mean"),
    Maximum_Salary=("Salary","max"),
Maximum_Experience=("Experience","max"),
Minimum_Experience=("Experience","min"),
)
                                                                       
```

```python
df.groupby("Department").agg(
    avg_salary=("Salary", "mean"),
    max_exp=("Experience", "max"),
    total_emp=("Employee", "count")
)
```
