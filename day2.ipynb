## Advance indexing

```python
import numpy as np
```

```python
q1=np.arange(1,11)
print(q1)
```

```python
a1=np.array([[12,13,14],[15,16,17]])
print(a1)
```

```python
a2 = ([10,20,30])
print(a2)
```

```python
a1+a2
```

```python
a3=q1.reshape(2,5)
a3
```

```python
a5 = np.array([[22],[27]])
a5
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
result=np.vstack((a,b))
print(result)
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
result=np.hstack((a,b))
print(result)
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
result=np.row_stack((a,b))
print(result)
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
result=np.column_stack((a,b))
print(result)
```
