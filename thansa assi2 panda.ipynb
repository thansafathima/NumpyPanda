## DATA CLEANING (REAL-WORLD SCENARIOS)

### 7. Identify missing values in each column

```python
print(df.isnull().sum())
```

### 8. Remove rows where "Rating" is missing

```python
df = df.dropna(subset=["Rating"])
```

### 9. Fill missing values in "Type" with the most frequent value (mode)

```python
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
```

### 10. Remove duplicate records from the dataset

```python
df = df.drop_duplicates()
```

### 11. Clean "Installs" (remove '+', ',' and non-numeric entries if present, convert to int)

```python
df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False)
)
```

### 12. Clean "Price" (remove '$' and convert to float)

```python
df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False)
)
```

### 13. Convert "Reviews" column to integer type

```python
df["Reviews"] = df["Reviews"].astype(int)
```

### 14. Convert "Last Updated" column to datetime format

```python
df['Last Updated'] = pd.to_datetime(
    df['Last Updated'],
    errors='coerce'
)
print(df['Last Updated'].dtype)
```

##  STRING OPERATIONS

### 15. Convert all app names to lowercase

```python
df["App"] = df["App"].str.lower()
```

### 16. Convert category names to uppercase

```python
df["Category"] = df["Category"].str.upper()
```

### 17. Split the "Genres" column into multiple parts (takes the first genre)

```python
df["Primary_Genre"] = df["Genres"].str.split(";").str[0]
```

### 18. Replace unwanted text patterns in "Category" (e.g., replacing underscores with spaces)

```python
df["Category"] = df["Category"].str.replace("_", " ")
```

## 5. FEATURE ENGINEERING

### 24. Create a new column "Revenue" using installs and price

```python
df["Revenue"] = df["Installs"] * df["Price"]
```

### 25. Categorize apps into rating groups (Low, Medium, High)

```python
df["Rating_Group"] = pd.cut(
    df["Rating"], bins=[0, 3.0, 4.2, 5.0], labels=["Low", "Medium", "High"]
)
```

### 26. Create a new column to flag high-install apps (>1M installs)

```python
df["Is_High_Install"] = df["Installs"].apply(lambda x: 1 if x > 1000000 else 0)
```

### 27. Convert "Type" column into numeric values (Free=0, Paid=1)

```python
df["Type_Numeric"] = df["Type"].map({"Free": 0, "Paid": 1})
```

### 39. Perform a left join and identify missing values created after the merge

```python
merged_data=pd.merge(apps_info,installs_info,on='App',how='left')

missing_values=merged_data[merged_data['Installs'].isnull()]

print("Missing values:")
print(missing_values)

print("Number of missing Installs:")
print(merged_data['Installs'].isnull().sum())
```
